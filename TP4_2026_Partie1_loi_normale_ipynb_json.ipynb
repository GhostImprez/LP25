# Loi de Gauss

L'objectif de ce Notebook est de présenter la distribution gaussienne (également connue sous le nom de distribution normale).

La distribution normale est assez omniprésente dans la vie. Quelques exemples : la taille des personnes, le poids des nouveau-nés à la naissance, etc. Il existe de nombreuses bibliothèques python que l'on peut utiliser pour générer des variables aléatoires, tracer et calculer les probabiltiés kiées à une distribution normale.

La fonction de densité de probabilité d'une distribution normale est donnée par la formule suivante

$$f(x) = \frac{1}{\sqrt{2\pi\sigma^2}}\exp\left(-\frac{(x - \mu)^2}{2\sigma^2}\right)$$

où le paramètre $\mu$ est la moyenne et $\sigma$ son écart-type. Le terme $\sqrt{2\pi\sigma^2}$ est connu sous le nom de [constante de normalisation](https://en.wikipedia.org/wiki/Normalizing_constant) et est
principalement utilisé pour réduire la fonction de probabilité à une fonction de densité de probabilité
avec une probabilité totale de un.

Écrivons maintenant une fonction qui renvoie une distribution gaussienne à partir de la moyenne et de l'écart-type. Nous allons écrire deux fonctions, *pdf_gaussian* et *pf_gaussian* où la première est une fonction de densité de probabilité (pdf) et la seconde est simplement une fonction de probabilité gaussienne.


````python
# import des librairies numpy et pyplot
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams["figure.figsize"] = (8,6)
````

````python
# Fonction qui calcul la loi de Gauss pour (mu=1, sigma) aux points x (x peut être un table)
def pdf_gaussian(sigma, mu, x):
    Z = np.sqrt(2.0*np.pi*sigma**2) # Constante de normalisation
    pdf_z = np.exp(-((x - mu)**2/(2.0*sigma**2)))/Z
    return pdf_z
````

````python
# echantillonnage de x et calcul de f(x) pour différentes valeurs de mu et sigma^2
x, dx = np.linspace(-1 ,3,300, retstep = True)
pdf_1 = pdf_gaussian(0.2, 1.0, x)
pdf_2 = pdf_gaussian(0.4, 1.0, x)
pdf_3 = pdf_gaussian(0.6, 1.0, x)
````

````python
# Tracé des densités pour différentes valeurs de sigma
plt.plot(x, pdf_1, label = '$mu=1$, $\sigma$ = 0.2')
plt.plot(x, pdf_2,label = '$mu=1$, $\sigma$ = 0.4')
plt.plot(x, pdf_3,label = '$mu=1$, $\sigma$ = 0.6')
plt.xlabel("x",fontsize = 14)
plt.ylabel("$f(x)$",fontsize = 14)
plt.legend()
plt.plot()
````

Dans le graphique ci-dessus, nous avons créé trois distributions gaussiennes avec une moyenne identique de $1$
et trois écarts types différents. On peut voir qu'en augmentant l'écart-type, les distributions commencent à s'étaler et que leur hauteur diminue également.

## Fonction de répartition

Que pouvons-nous utiliser pour justifier que nos données suivent une distribution normale ? La réponse est la fonction de répartition. La fonction de répartition de la distribution normale et la distribution de la Taille doivent être semblables. Nous pouvons donc superposer la fonction de répartition de la distribution normale et la fonction de répartition de la Tailles, et voir si les deux CDF se superposent bien.

    



````python
fig, ax = plt.subplots()
num_bins = 100

# Histogramme des tailles
counts, bins = np.histogram(population_height,
                            np.linspace(min(population_height),max(population_height)),
                            num_bins)
# cumul des effectifs (en proportion)                            )
pmf = counts/sum(counts)
cdf = np.cumsum(pmf)
                            # Représentation
ax.scatter(bins[:-1], cdf, c='b', marker="+", label="Distribution cumulée des Tailles")

# CDF of Normal Distribution
mean, std = np.mean(population_height), np.std(population_height)
x = np.linspace(mean - 4*std, mean + 4*std, 100)
cdf = norm.cdf(x, mean, std)
ax.plot(x, cdf, c='g', label="Fonction de répartition de Gauss")
ax.legend()

plt.xlabel("Tailles (Inches)")
plt.ylabel("Fonction de répartition des Tailles \n Fonction de répartition de la loi Normale")
plt.show()
````

```
result = 1.-norm.cdf(67.12, mean, std)
```
